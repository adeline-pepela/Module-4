```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt
import numpy as np
```

**1. Load dataset**

```python
(X_train, y_train), (X_test,y_test) = datasets.cifar10.load_data()
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
# Reshape the data
y_train  = y_train.reshape(-1,)
y_test   = y_test.reshape(-1,)
```

```python
# Label Classes
classes = ["airplane","automobile","bird","cat","deer","dog","frog","horse","ship","truck"]
```

```python
# Normalizing the training data
X_train = X_train / 255.0
X_test  = X_test / 255.0
```

```python
ann = models.Sequential([
        layers.Flatten(input_shape=(32,32,3)), #converts a 3D input (32x32x3) into a 1D vector
        layers.Dense(3000, activation='relu'), # First dense layour
        layers.Dense(1000, activation='relu'),
        layers.Dense(10, activation='softmax') # Output Layer
    ])

# Compiling the Model
ann.compile(optimizer='SGD',
              loss='sparse_categorical_crossentropy', #loss function is used for multi-class classification
              metrics=['accuracy'])

ann.fit(X_train, y_train, epochs=5)
```

```python
ann.evaluate(X_test,y_test) # Evaluate the model
```

```python
from sklearn.metrics import confusion_matrix , classification_report
import numpy as np

y_pred          = ann.predict(X_test)
y_pred_classes = [np.argmax(element) for element in y_pred]

print("Classification Report: \n", classification_report(y_test, y_pred_classes))
```

**Now let us build a convolutional neural network to train our images**

```python
cnn = models.Sequential([

  #  Performs 2D convolution to detect features such as edges or textures in the image
    layers.Conv2D( filters=32,           # 32 different filters will be learned in this layer
                   kernel_size=(3, 3),   # The filter size (3x3) that slides over the input to extract features
                   activation='relu',    # adds non-linearity, ensuring the network can model complex relationships
                   input_shape=(32, 32, 3)
                   ),
    layers.MaxPooling2D((2, 2)),  # Reduces the spatial dimensions (width and height) of the input by taking the maximum value from each 2x2 block

   # deeper convolutional layer with 64 filters to  detect more complex patterns compared to the first laye
    layers.Conv2D(filters=64,
                  kernel_size=(3, 3),
                  activation='relu'),

    layers.MaxPooling2D((2, 2)),# max-pooling layer further reduces the spatial size of the feature maps.
    layers.Flatten(),

    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])
```

```python
# Compile model

cnn.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
# Train the model
cnn.fit(X_train, y_train, epochs=10)
```

```python
# Evaluate
cnn.evaluate(X_test,y_test)
```

**Use of Transfer Learning**

```python
from tensorflow.keras.applications import ResNet50
from tensorflow.keras import layers, models
from tensorflow.keras.optimizers import SGD
```

```python

# Load the ResNet50 model with pre-trained weights, excluding the top layers (used for ImageNet classification)
base_model = ResNet50(weights='imagenet',
                      include_top=False,
                      input_shape=(32, 32, 3))  # Adjust input shape to your dataset (CIFAR-10 format)

# Freeze the base model layers to prevent them from being retrained
base_model.trainable = False

# Build your new model on top of ResNet
model = models.Sequential([
    base_model,        # Pre-trained ResNet as the feature extractor
    layers.Flatten(),

    # Add your custom Dense layers for classification
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.5),  # Prevent overfitting with Dropout
    layers.Dense(10, activation='softmax')  # Output layer for 10 classes
])

# Compile the model
model.compile(optimizer=SGD(learning_rate=0.001, momentum=0.9),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

```

```python
# Train the model on your dataset
history = model.fit(X_train, y_train, epochs=5)
```

```python
model.evaluate(X_test,y_test)
```
